```python
import pandas as pd
```

```python
dataset = pd.read_csv("insurance_pre.csv")
dataset
```

```python
dataset = pd.get_dummies(dataset, drop_first=True)
```

```python
independent = dataset[['age', 'bmi', 'children', 'sex_male', 'smoker_yes']]
dependent = dataset[["charges"]]
```

```python
from sklearn.model_selection import train_test_split
x_train_non_standardized, x_test_non_standardized, y_train, y_test = train_test_split(independent, dependent, test_size=0.30, random_state=0)
```

```python
from sklearn.tree import DecisionTreeRegressor
regressor_squared_error_best=DecisionTreeRegressor(criterion='squared_error', splitter='best')
regressor_squared_error_best=regressor_squared_error_best.fit(x_train_non_standardized, y_train)
```

```python
import matplotlib.pyplot as plt
from sklearn import tree
tree.plot_tree(regressor_squared_error_best)
plt.show()
```

```python
y_pred_squared_best=regressor_squared_error_best.predict(x_test_non_standardized)
```

```python
from sklearn.metrics import r2_score
r_score_squared_best = r2_score(y_test,y_pred_squared_best)
r_score_squared_best
```

```python
from sklearn.tree import DecisionTreeRegressor
regressor_squared_error_random=DecisionTreeRegressor(criterion='squared_error', splitter='random')
regressor_squared_error_random=regressor_squared_error_random.fit(x_train_non_standardized, y_train)
```
