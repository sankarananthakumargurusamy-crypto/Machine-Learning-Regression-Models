```python
import pandas as pd
```

```python
dataset = pd.read_csv("50_Startups.csv")
```

```python
dataset
```

```python
dataset = pd.get_dummies(dataset, drop_first=True)
```

```python
dataset.columns
```

```python
independent = dataset[['R&D Spend', 'Administration', 'Marketing Spend', 'Profit',
       'State_Florida', 'State_New York']]
```

```python
dependent = dataset[["Profit"]]
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(independent, dependent, test_size=0.30, random_state=0)
```

```python
from sklearn.tree import DecisionTreeRegressor
regressor=DecisionTreeRegressor()
regressor=regressor.fit(x_train, y_train)
```

```python
import matplotlib.pyplot as plt
from sklearn import tree
tree.plot_tree(regressor)
plt.show()
```

```python
y_pred=regressor.predict(x_test)
```

```python
from sklearn.metrics import r2_score
r_score = r2_score(y_test,y_pred)
```

```python
r_score
```

```python
from sklearn.tree import DecisionTreeRegressor
regressor_friedman=DecisionTreeRegressor(criterion='friedman_mse', splitter='best')
regressor_friedman=regressor_friedman.fit(x_train, y_train)
```

```python
import matplotlib.pyplot as plt
from sklearn import tree
tree.plot_tree(regressor_friedman)
plt.show()
```

```python
y_pred_friedman=regressor_friedman.predict(x_test)
```

```python
from sklearn.metrics import r2_score
r_score_friedman = r2_score(y_test,y_pred_friedman)
```

```python
r_score_friedman
```
